```python
## write the content into file line by line  -> writelines()

lines = ['first_lline\n','second_line\n','third_line\n']
with open('example.txt','w') as data :
    data.writelines(lines)
```

```python
# r,w,a

# mode --> write the content into the binary files -> wb
# mode --> Read the content from the binary file -> rb

data = b'\x00\x01\x02\x03\x04'
with open('example.txt','wb') as binary_data:
    binary_data.write(data)
```
