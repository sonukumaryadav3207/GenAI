```python
## open() -> open file
## first paramte -> file name
## second parameter -> mode
## mode -> it decides the way in which wwe are writing this file

f=open('example.txt','r')
content = f.read()
print(content)
```

```python
## write() -> eriting the content into the file
## 1, overerites then old content write()
## add data at the end of old file append()

f=open('example.txt','w')
f.write('Hello i am sonu yadav do you know me')
f.close()
```

```python
## append ()

f= open('example.txt','a')
f.write("\n We are learning Gen AI")
f.close()
```

#### with -> it will be used for all opening the file as a objet  and there is no requirement for closeing the file as well

```python
with open('example.txt','r') as f:
    content_data = f.read()
    print(content_data)
```

```python
with open('example.txt','w') as file:
    file.write("\n we are exploreing with command in pythin file handling")
```
