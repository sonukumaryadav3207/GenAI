## Shallow Copy is being used for creating a copy of the original dictionary in a way such that the changes that we are going to change in original dictionary wont be getting reflected into our copied dictionary and  shallow coly is being done by copy()

```python
dictionary_11={"name":"Amit","a":23,"b":25,"f":29}
```

```python
dictionary_21=dictionary_11
```

```python
dictionary_21
```

```python
dictionary_11['a']=46
```

```python
dictionary_11
```

```python
dictionary_21
```

```python
dictionary_3=dictionary_11.copy()
```

```python

dictionary_3
```

```python
nested_dictionary = {
    "student1":{"Name":"amit","marks":85},
    "student2":{"Name":"Sonu","marks":88}
                     }
```

```python
nested_dictionary
```

```python
for outer_key,outer_value in nested_dictionary.items():
    print(f"Outer Key is :{outer_key} and outer value is :{outer_value}")
    for key_1,value_1 in outer_value.items():
        print(f"Inner Key is : {key_1} and Inner value is : {value_1}")
```
