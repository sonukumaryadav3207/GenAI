# Map

## Map function applies a givrn function operation to All items inan iterable or a List and  returns a Map object

## Map function is particulary useful for Transforming the data or applying  an operation on ALL items of the iterable

### syntax for map function 

### list(map (function,iterable)) 

```python
def square(list_items):
    empty_list =[]
    for item in list_items:
        squared_item=item**2
        empty_list.append(squared_item)
    print(empty_list)
square([1,4,5,3,8,6,9])
```

```python
print(list(map(lambda x:x**2,[1,4,5,3,8,6,9])))
```

```python
num1= [1,2,3,4,5]
num2=[1,2,3,4,5]

print(list(map(lambda x,y:x+y,num1,num2)))
```

```python
# map () to convert all string value items in an interable into an integer
map_value=list(map(int,['1','2','3','4','5']))
print(map_value)
```

```python
emp_name = ['amit','ravi','soali','kunal','usha','radha','sonu']
print(list(map(str.upper,emp_name)))
```

```python
emp=[
    {'Name':'Amit' ,'Age':34,'qualification':'MCA'},
    {'Name':'Sumit' ,'Age':23,'qualification':'BCA'},
    {'Name':'Harsit' ,'Age':14,'qualification':'10th'},
    {'Name':'Amar' ,'Age':32,'qualification':'MTECH'}
]

def age(list_item):
    return list_item['Age']

print(list(map(age,emp)))
```
