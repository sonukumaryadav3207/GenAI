```python
## to create a function that converts temprature from celcius to fahrenhit or fahrenhrit to celcius

## if temprature is in celcius , then celcius to fahrenheit conversion is :

''' 1 deg celcius = 9/5 +32 
     45 deg C = 45*9/5 +32 fahrenheit

     1 fahrenheit celcius =5/9 - 32
      45 def F = 45-32 * 5/9 deg c
 '''
def tem_change(temp_value,unit):
    if unit == 'C':
        temprature =  temp_value * 9/5 +32
        print(f'The Temprature for {temp_value} deg celcius in fahrenheit is : ', str(temprature) + ' F')
    elif unit == 'F':
        temprature = round((temp_value - 32) * 5/9 , 2)
        print(f'The temprature for {temp_value} deg fahrenheit in celcius is :' , str(temprature) + 'C')
    else:
        print("Invalid")
tem_change(45,'F')
```

```python
## strong possword checker
''' Strong password  -->
    1. length is greater than 8 characters
    2. password must consist atleast a lowercase chareacter
    3. password nust be consist of atleast a digit
    4. password msut contain  atleast a uppercase
    5. password must contain a special character

Weak Password -->
    If a password doesnt match any above condition then password as weak password '''

def  password_strength(password):
    if len(password) < 8:
        print(f"The enterd {password} is a weak password")
    elif not any (char.islower() for char in password):
       print(f"Ther enter password {password} is a weak password")
    elif not any(char.isupper() for char in password):
       print(f"Ther enterred password {password} is a weak password")
    elif not any(char.isdigit() for char in password):
       print(f"Ther enterred password {password} is a weak password")
    elif not any(char for char in '!,@,$,#,%,^,&,<,*,(,),_,-'):
       print(f"Ther enterred password {password} is a weak password")
    else:
       print(f"Ther enterred password {password} is a Strong password")
```

```python
def palindrome(string):
    string_value =  string.lower().replace(" ","")
    if string_value == string_value[:: -1]:
        print("Palindrome")
    else:
        print("Not palindrome")


palindrome("A ada  a")
```
