# Lamda function

## Lamda functions are the One Liner Functions define by the Lamda keyword

## Syntax for Lamda Function

## lamda arguments:expression

## lamda --> keyword which simply let the python interpreter know that the current function is our Lamda Function

## Argumrnts --> Input values for our function 

## Expression --> Logic (Funvtion)

```python
##  function which calculate the square of the numbers

def square(num):
    return num ** 2
```

```python
print(square(5))
```

```python
lambda_function = lambda x:x**2
type(lambda_function)
lambda_function(5)
```

```python
## lambda function to calculate the sum of the numbers

sum= lambda x,y:x+y
print(sum(5,6))
```

```python
product = lambda a,b,c,d:a*b*c*d
print(product(2,3,4,5))
```
