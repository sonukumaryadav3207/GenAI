# filter function


```python
def even_function(value):
    if value%2 == 0:
        return True
```

```python
list_item = [1,3,4,3,5,6,7,8,9,65,6,76,78,99,876,56]
print(list(filter(even_function,list_item)))
```

```python
print(list(filter(lambda x:x%2==0,list_item)))
```

```python
print(list(filter(lambda x:x%2==0 and x>10 , list_item)))
```

```python
emp=[
    {"name":"Amit","Age":12,"qualification":"MCA"},
    {"name":"sumit","Age":22,"qualification":"BCA"},
    {"name":"harmit","Age":21,"qualification":"DCA"},
]

def age(emp):
    if emp['Age'] > 15:
        return True

print(list(filter(age,emp)))
```
