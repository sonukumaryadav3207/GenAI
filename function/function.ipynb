# Recursive function

## argument = function calling()

## parameter =  functuion defination()

## function calling it self again and again

```python
def factorial(num):
    """  this is the translator """
    if num == 0 or num == 1:
        print(f"The factorial of {num} is : 1")

    else:
        factorial = 1
        for i in range(1,num+1):
            factorial=factorial*i
        print(f"The factorial of num is : {factorial}")
```

```python
factorial(5)
```

```python
def factorial_recursive(num):
    if num == 0 or num ==1:
        return 1
    else:
        return num * factorial_recursive(num-1)
```

```python
factorial_recursive(5)
```

```python
def factorial(num):
    if num == 1 or num == 0:
        print(f"factorial of {num} is 1")
    else:
        factorial = 1
        for i in range(1,num+1):
            factorial = factorial * i
        print(f"factorial of {num} is {factorial}")

factorial(0)        
```

# kayword Arguments -> takes the input as the keyword values and irrrespective of any number of keyword values you are going to pass it wont matter

```python
def data(**user_data):
    for key,value in user_data.items():
        print(f'{key}:{value}')
```

```python
data(name="Amity",age=25,gender="m")
```

## want to pass keyword value pairs as your input along with positional argument as well

```python
def variable(*postionalm,**keywords):
    for i in postionalm:
        print(f'positional vallues are ',i)
    for key,value in keywords.items():
        print(f'key is :{key} and value is : {value}')
```

```python
variable(1,2,3,4,5,6,name="Amity",age=25,gender="m")
```
