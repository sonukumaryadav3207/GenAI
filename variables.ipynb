# Varibale Name Convention

```python
''' 1. Variable name must be Descriptive : 
    Name = 12 here Name is not descriptive variable name.
    Age = "Sonu" age is not descriptive variable name.

    2. Variable name should not be a keyword :
    for = 12 here for is a keyword in python so it can not be used as

    3. Variable name should not start with a number :
    1name = 12 here 1name is not a valid variable name because it starts with a number.

    4. Variable name should not contain special characters :
    name@ = 12 here name@ is not a valid variable name because it contains special character @.

    They must start with a letter or underscore and can contain letters, numbers, and underscores.

'''
```

## Input function only takes the input in a string format until and unless we are not going to explicitly take the input in integer or other format

```python
Age = input("Enter your age :")
print(type(Age))
```

```python
Age  = int(input("Enter your age :"))
print(type(Age))
```

```python

## building a simple calculator using python
value1 = int(input("Enter first value :"))
value2 = int(input("Enter second value :"))
print("Addition of two values is:", value1 + value2);
print("Subtraction of two values is:", value1 - value2);
print("Multiplication of two values is:", value1 * value2);
print("Division of two values is:", value1 / value2);
print("Modulus of two values is:", value1 % value2);
print("Exponent of two values is:", value1 ** value2);
print("Floor division of two values is:", value1 // value2);
print("Greater value is:", max(value1, value2));
print("Smaller value is:", min(value1, value2));
print("Average of two values is:", (value1 + value2) / 2);
```
