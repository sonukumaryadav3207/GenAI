# Inherirtance 

#### it is a fundamental concept in object oriented programming because of which the child is able to inherit the propertiess OR attributes and functionalities of the parent class

### Just like a child is going to inherit the properties from parent is going to inherit the properties from the grandparent

#### In python typically we get to see 3 types of Inhertance :

### 1. Single Inheritance
### 2. Multilevel Inheritanse
### 3. Multiple Inheruitance

#### Single Inhertinace --> The Fetures or the attributes from one parent class will be getting inherted into ome chlid class


```python
class Avenger:
    def __init__(self,name,power):
        self.name = name
        self.power=power
    
    def introduction(self):
        print(f"Hello My name is {self.name}")

class Ironman(Avenger):
    def __init__(self,name,power,weapon):
        super().__init__(name,power)
        self.weapon=weapon

    def weapon_deatails(self):
        print(f"I use {self.weapon}")
```

```python
class Grandparent:
    def __init__(self, nationality,baldness,anger):
        self.nationality=nationality
        self.baldness=baldness
        self.anger=anger

    def introduction(self):
        print(f"Hello my nationality is : {self.nationality} and about my anger_issue is {self.anger} and for baladness {self.baldness}")
```

```python
class Parent(Grandparent):
    def __init__(self,nationality,baldness,anger,intelligence):
        super().__init__(nationality,baldness,anger)
        self.intelligence= intelligence

    def intelligence(self):
        print(f"you know my intelligence level {self.intelligence}")
```

```python
class Chlid(Parent):
    def __init__(self,nationality,baldness,anger,intelligence,charm):
        super().__init__(nationality,baldness,anger,intelligence)
        self.charm=charm

def charm_details(self):
    print(f"ABout my charm i am {self.charm}")
```
