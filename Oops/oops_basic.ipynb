```python
class Avenger:
    def __init__(self,name,power,weapon,nationality="America"):
        self .name = name
        self.power=power
        self.weapon=weapon
        self.nationality=nationality

    def introduction(self):
        print(f"Hello my name is {self.name}")
    def carrying_weapon(self):
        print(f"I am carrying weapon {self.weapon}")
```

```python
iron_man=Avenger("Iron Man","Genius","Repulsor beams")
thor = Avenger("Thor", "God", "Hammer", "Asgard")

iron_man.introduction()
iron_man.carrying_weapon()
iron_man.nationality
thor.introduction()
thor.carrying_weapon()
thor.nationality
```

#### types of method

#### 1. Instance Method ---> Instane Methods operate an object attributes

#### 2. Class Method--- > It  is the method of the whole class or the whole class or the blueprint which will remain the same for all of the objects then sunc kind of method is being reffered as a classmethod

## in class method , these methods work with the class level attributes .

#### 3. Static Method ---> do not use self or cls

```python
## instamce method

class Avengers:
    def __init__(self,name,power):
        self.name=name
        self.power=power

    def show_details(self):
        print(f"My name is {self.name} and i am having power {self.power}")

iron_man = Avengers("Iron Man","Genius")
iron_man.show_details()
        
```

```python
class Endgame:
    origin = "Marvel"

    def __init__(self, name, power):
        self.name = name
        self.power = power

    @classmethod
    def origin_info(cls):
        print(f"My origin is {cls.origin}")

    def self_info(self):
        print(f"My name is {self.name} and my power is {self.power}")


Iron_man = Endgame("Iron Man", "Genius")

Iron_man.origin_info()
Iron_man.self_info()
```
