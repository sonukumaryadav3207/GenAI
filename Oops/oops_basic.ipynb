```python
## instamce method

class Avengers:
    def __init__(self,name,power):
        self.name=name
        self.power=power

    def show_details(self):
        print(f"My name is {self.name} and i am having power {self.power}")

iron_man = Avengers("Iron Man","Genius")
iron_man.show_details()
        
```

```python
class Endgame:
    origin = "Marvel"

    def __init__(self, name, power):
        self.name = name
        self.power = power

    @classmethod
    def origin_info(cls):
        print(f"My origin is {cls.origin}")

    def self_info(self):
        print(f"My name is {self.name} and my power is {self.power}")


Iron_man = Endgame("Iron Man", "Genius")

Iron_man.origin_info()
Iron_man.self_info()
```

```python
Iron_man = Endgame("Iron Man" ,"Genius")
```

```python
Iron_man.origin_info()
```
