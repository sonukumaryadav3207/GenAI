# Opreator Overloading

## Opreator Overloading allows us to define how operators(+,-,/*,etc) work with user defined objects .

## why is Operator Overloading useful ?

## Makes custom objects nehave like built in types (number,string,lists)
## Improves code readability bu using natural syntax(object1+object2) instead od object1.add (object2)



## Operator --> +  ---> Method name  -- > __add__(self,other)
## Operator --> -   ---> Method name --> __sub__(self,other)
## Operator --> *  ---> Method name --> __mul__(self,other)
## Operator --> /  ---> Method name --> __truediv__(self,other)


```python
class Avenger:
    def __init__(self,name,power):
        self.name=name
        self.power=power

    
    def __add__(self,other):
        return Avenger(
            self.name + " & " +  other.name , 
            self.power + other.power
        )
    

    def __sub__(self,other):
        return Avenger(
            self.name , 
            self.power  - other.power
        )

    def __mul__(self,other):
        return Avenger(
            self.name , 
            self.power * other.power
        )

    def __truediv__(self, other):
        return Avenger(
            self.name,
            self.power/other.power
        )
    
    def __str__(self):
        return f"{self.name} with Power {self.power} "
```

```python
thor=Avenger("Thor",100)
hulk=Avenger("Hulk",150)
```

```python
team = thor  - hulk
```

```python
print(team)
```

```python
class AIModel:
    def __init__(self,name,accuracy):
        self.name=name
        self.accuracy=accuracy

    def  __gt__(self, other):
        return self.accuracy > other.accuracy

    def __lt__(self,other):
        return self.accuracy < other.accuracy

    def  __eq__(self, other):
        return self.accuracy == other.accuracy
    
```

```python
gpt4=AIModel("GPT-4",98)
bard=AIModel("Google Bard",96)
```

```python
print(gpt4>bard)
```

```python
print(gpt4==bard)
```
