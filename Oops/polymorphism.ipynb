# Polymorphism

## Polymorphism means Many forms in Greek.


## In oops , Polymorphism allows object of different classes as if they are the object of Different classes as if they are the object of the same class & It enable reusability and makes program more flexible .

## Type of Polymorphism

#### 1. Method Overriding (Runtime Polymorphism)
#### 2. Method overloading(Compiletime Overloading)
#### 3. Operator Overloading

Method Overriding -- > A child class provides a specific implementaion of a method that has already been defined into the parent class.

This allows dynamic selection at Runtime.


```python
class Grandparent:

    def speak(self):
        print("Heloo")

class Parent(Grandparent):
    def speak(self):
        print("Namaste")

class Child(Grandparent):
    def speak(self):
        print("bon joure")
```

```python
def attribute_provider(objects):
    print(objects.speak())
```

```python
granda = Grandparent()
```

```python
print(attribute_provider(granda))
```
