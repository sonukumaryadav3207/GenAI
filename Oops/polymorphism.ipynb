# Polymorphism

## Polymorphism means Many forms in Greek.


## In oops , Polymorphism allows object of different classes as if they are the object of Different classes as if they are the object of the same class & It enable reusability and makes program more flexible .

## Type of Polymorphism

#### 1. Method Overriding (Runtime Polymorphism)
#### 2. Method overloading(Compiletime Overloading)
#### 3. Operator Overloading

Method Overriding -- > A child class provides a specific implementaion of a method that has already been defined into the parent class.

This allows dynamic selection at Runtime.


```python
class Grandparent:

    def speak(self):
        print("Heloo")

class Parent(Grandparent):
    def speak(self):
        print("Namaste")

class Child(Grandparent):
    def speak(self):
        print("bon joure")
```

```python
def attribute_provider(objects):
    return objects.speak()
```

```python
grandpa = Grandparent()
```

```python
print(attribute_provider(grandpa))
```

```python
parent = Parent()
```

```python
attribute_provider(parent)
```

```python
child = Child()
```

```python
attribute_provider(child)
```

```python
class Shape:
    def area(self):
        return "This mrthod returns the area of the shape"
    
```

```python
class Rectange(Shape):
    def __init__(self,width,height):
        self.width=width
        self.height=height
        
    def area(self):
        return self.width*self.height

class Circle(Shape):
    def __init__(self,radius):
        self.radius=radius

    def area(self):
        return 3.14*self.radius*self.radius
```
