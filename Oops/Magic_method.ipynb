# Magic Method

## Magic Method are also called as Dunder Methods becasuse they have double underscore __ brfore and after their name and they are specail methods in python that allow us to customize how objects behave

## These metods can help us define :

## 1. Object Intializatio (__init__)
## 2. String Represtation(__str__,__repr__)
## 3. Airthmetic Operation(__add__,__sub__,etc)
## 4. Comapre Operations (__eq__,__it__)
## 5. Length , Indexing ,itreation(__len__,__getitem__,etc)

## Why to use Magic Methods

## print(object) => __str__ --> what is returned when print(object) is used 

## Airthmetic operration can be done among the objects by using Magic Methods

```python
## __add__

class Avenger:
    def __init__(self,name,strength):
        self.name=name
        self.strength=strength
    
    def __add__(self,other):
        return Avenger(self.name + " & " + other.name , self.strength + other.strength)
    
    def __str__(self):
        return f"{self.name} with combined strength {self.strength}"


ironman = Avenger('Ironmon',80)
thor=Avenger("Thor",100) 

team = ironman + thor
print(team)
```
