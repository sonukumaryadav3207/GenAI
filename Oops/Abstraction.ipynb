# Abstraction

## Abstraction is the concept of Hiding the implementation details and only showing the necessary features of the object

## This helps in reducing programing complexity and efforts

## Abstraction allows us to work with an object with what an object does rather than how it does it

### How can we achieve abstration in python 

## 1. Abstract Class - >  A class  that can not be instantiated or that means for an abstract class we can not create the object og this class .

## 2. Abstract Method - > Abstract method just because it cant accessed by the object of the abstract class itself why we have to call this abstract method in our child class and the  implemntation of the abstract method is not  provided inside the abstract class  in the child class

## The abstract method id being defined by @abstractmethod

```python
from abc import ABC, abstractmethod

class Vehical(ABC):
    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def stop(self):
        pass

class Car(Vehical):

    def start(self):
        print("The car starts with an ignition key")
    
    def stop(self):
        print("The car stops with break")

class ElectricScooter(Vehical):

    def start(self):
        print("Electic scooter starts with power button")

    def stop(self):
        print("Elecctric scotter stops with electric break")
```

```python
car = Car()
```

```python
car.start()
```
