## Inhertance 

#### Inheritance is a Fundamental concept in Object Oriented Programming because of which the child class is able to inherit the properties or Attributes and functionalities of the parent class

#### Just like a child is going to inherit the properties from parent and parent is going to inherti the properties from the grandparent and defiinitiely in this hierarchy the grandparent as well

#### In python typically we get to see 3 Types of inherttance :

#### 1. Single Inheritance
#### 2. Multilevel Inheritance 
#### 3. Multiple Inheritancce

#### 1. Single Inhertitance --> The Features or the Attributes form one parent class will be getting inhertited into one child class.



```python
class Avenger:

    def __init__(self,name,power):
        self.name=name
        self.power=power

    def introduction(self):
        print(f"Hello my name is {self.name}")

class IronMan(Avenger):
    def __init__(self, name, power,weapon):
        super().__init__(name, power)
        self.weapon=weapon

    def weapon_details(self):
        print(f"I use {self.weapon}")
```

```python
Iron=IronMan("Iron Man","Genius","Repulsor beams")
```

```python
Iron.name
```

```python
Iron.weapon_details()
```

```python
Iron.introduction()
```

#### 2. Multilevel Inhertance

#### The properties or method or attribute of the grandparent calss can be inhertited by the parent class and then the properties of the parent class  will be inherited by the child class. In a proper heirarchy the Inheritance process is going to work for us.

```python
class Granparent:
    def __init__(self,nationality,baldness,anger_issue):
        self.nationality=nationality
        self.baldness=baldness
        self.anger_issue=anger_issue

    def  intoduction(self):
        print(f"Hello my nationality is : {self.nationality} and about my anger issue if you want to ask about then the answer is {self.anger_issue}")
        
```

```python
class Parent(Granparent):
    def __init__(self, nationality,baldness,anger_issue,intelligence):
        super().__init__(nationality,baldness,anger_issue)
        self.intelligence=intelligence

    def intelligence_details(self):
        print(f"About my intelligence if you guys want to know that if  i am intelligent or not then the answer is {self.intelligence}")
```

```python
class Child(Parent):
    def __init__(self,nationality,baldness,anger_issue,intelligence,charm):
        super().__init__( nationality,baldness, anger_issue,intelligence)
        self.charm =charm

    def charm_details(self):
        print(f"About my charm if you guys want to know that whether if i am charming or not thne the answer is {self.charm} ")
```

```python
parent_1 =Parent("Indian","Yes","No","Yes")
```

```python
parent_1.intoduction()
```

```python
child_1=Child("Indian","No","Yes","Yes","Yes")
```

```python
child_1.intoduction()
```

```python
child_1.intelligence_details()
```

```python
child_1.nationality="Sikh"
```
