# Encapsulation

## Encapsulation is way of wrapping data (variable and method) inside a capsule or a class as a single unit.

### It restrict the direct access to some of the objects comonents's which is a means of preventing accidental inference and misuse of the data


#### Key concept of Encapsulation :

## 1. Public Members
## 2. protected Members
## 3. Private Member
## getter and setter methods

## public member --> Accessible from Anywhere or Attributes are accessible outside of the class and these members are accessible by the derived class

## public member attribute is directly decalred 

```python
class Ironman:
    def __init__(self,name,power,weapon):
        self.name=name
        self.power=power
        self.weapon=weapon

    def intro(self):
        print(f"Hello  i am {self.name} and i have {self.weapon}")
```

## private member  --> A private member is not accessible outside the class and not even in the derived class and in oreder to have the access over these private members or to bring some modificatiomn on these members we use getter and setter method

## Getter method = it will be helping  us out with accessing the private data

```python
class Caption:
    def __init__(self,name,address):
        self.name=name
        self.__address=address

    def accessing_add(self):
        print (f"My address is {self.__address}")
```

```python
caption=Caption("Caption","new york")
```

```python
caption.name
```

```python
caption.__address
```

```python
dir(caption)
```

```python
caption.accessing_add()
```

## settert Method = > setter method is being used for modifying private data

```python
class Animal:
    def __init__(self,name,sound):
        self.name=name
        self.__sound=sound

    def sound_info(self):
        print(f"The animal is goinfg to sound as : {self.__sound}")

    def sound_modify(self,new_sound):
        self.__sound=new_sound
        print(f"New sound for the animal is : {self.__sound}")
```

```python
animal = Animal("Bruno","Woof")
```

```python
animal.name
```

```python
animal.sound
```

```python
animal.sound_info()
```

```python
animal.sound_modify("meaw")
```

## protyected data is the data is not accessible outside the class but the protected data is going to be accessible by the derived class or child class

## For having the protected member you can mention your protectwed member as _attribute or by using a single underscore

```python
class Grandparent:
    def __init__(self,name,locker_key):
        self.name=name
        self._locker_key=locker_key

    ## getter method
    def locker_info(self):
        print(f"The locker kry inof is {self.locker_key}")
```

```python
class Parent(Grandparent):
    def __init__(self,name,locker_key):
        super().__init__(name,locker_key)

    def child_reveals(self):
        print(f"the Locker Key information is : {self._locker_key}")
        
```

```python
grandpa=Grandparent("Amit","1234")
```

```python
grandpa.name
```

```python
grandpa.locker_key
```

```python
dir(grandpa)
```

```python
parent = Parent("Amit","1234")
```

```python
parent.locker_key
```

```python
parent.child_reveals()
```

# Bank Account System

```python
class Bankaccount:
    def __init__(self,account_holder,balance):
        self.account_holder=account_holder
        self.__balance =balance

    def  get_balance(self):
        return self.__balance

    def deposit(self,amount):
        if amount > 0:
            self.__balance += amount
            return f"Deposite amount is :{amount} . Total Balance is : {self.__balance}"

        else :
            return "Deposite Amount must be Positive"


    def withdraw(self,amount):
        if 0<amount<=self.__balance:
            self.__balance -= amount
            return f"withdrawal of {amount} is sucessful. Balance is  : {self.get_balance()} "
        else:
            return "Insufficient fund in the account"
        
```

```python
account=Bankaccount("Amit",1000)
```

```python
account.get_balance()
```

```python

account.account_holder
```

```python
account.withdraw(100)
```

```python
account.get_balance()
```
