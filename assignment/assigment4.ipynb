# Task 1

```python
sales = [1200, 450, 980, 3000]

file = open("sales_data.txt", "w")

for sale in sales:
    file.write(str(sale) + "\n")

file.close()

file = open("sales_data.txt", "r")
content = file.read()

print(content)

file.close()
```

# Task 2

```python
file = open("sales_data.txt", "r")

# read()
content = file.read()
print("Using read():")
print(content)

file.close()


file = open("sales_data.txt", "r")

# readline()
line = file.readline()
print("Using readline():")
print(line)

file.close()


file = open("sales_data.txt", "r")

# readlines()
lines = file.readlines()

sales = []

for line in lines:
    sales.append(int(line.strip()))

print("Using readlines():")
print(sales)

file.close()
```
